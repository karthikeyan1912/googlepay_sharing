```python
import numpy as np
from prettytable import PrettyTable
```

```python
# Friends
friends = ["Alice", "Bob", "Carol", "David"]

# Store expenses
expenses = []
```

```python
# Add Expense
def add_expense():
    print("\nFriends:", friends)

    payer = input("Who paid? ")

    if payer not in friends:
        print("Invalid payer!")
        return

    beneficiaries_input = input(
        "Enter beneficiaries separated by comma: "
    )

    beneficiaries = [
        name.strip()
        for name in beneficiaries_input.split(",")
    ]

    # Check beneficiaries
    for person in beneficiaries:
        if person not in friends:
            print("Invalid beneficiary:", person)
            return

    description = input("Enter expense description: ")

    amount = float(input("Enter total amount: ₹"))

    # Equal split
    share = amount / len(beneficiaries)

    expenses.append({
        "payer": payer,
        "description": description,
        "amount": amount,
        "beneficiaries": beneficiaries,
        "share": share
    })

    print("\nExpense added successfully!")
    print("Description :", description)
    print("Paid by     :", payer)
    print(f"Amount      : ₹{amount:.2f}")
    print(f"Each share  : ₹{share:.2f}")
```

```python
def calculate_settlements():

    balances = {
        person: 0.0
        for person in friends
    }

    for expense in expenses:

        payer = expense["payer"]
        amount = expense["amount"]
        beneficiaries = expense["beneficiaries"]
        share = expense["share"]

        # Payer paid the total amount
        balances[payer] += amount

        # Each beneficiary owes their share
        for person in beneficiaries:
            balances[person] -= share

    return balances
```

```python
# Display Settlements
def display_settlements():

    balances = calculate_settlements()

    table = PrettyTable()

    table.field_names = [
        "Friend",
        "Balance",
        "Status"
    ]

    for person in friends:

        balance = balances[person]

        if balance > 0.01:
            status = "Should Receive"

        elif balance < -0.01:
            status = "Owes"

        else:
            status = "Settled"

        table.add_row([
            person,
            f"₹{abs(balance):.2f}",
            status
        ])

    print("\nFinal Settlements")
    print(table)
```

```python
# Suggest Payments
def suggest_payment():

    balances = calculate_settlements()

    creditors = [
        [person, amount]
        for person, amount in balances.items()
        if amount > 0.01
    ]

    debtors = [
        [person, -amount]
        for person, amount in balances.items()
        if amount < -0.01
    ]

    transactions = []

    i = 0
    j = 0

    while i < len(debtors) and j < len(creditors):

        debtor = debtors[i]
        creditor = creditors[j]

        payment = min(
            debtor[1],
            creditor[1]
        )

        transactions.append(
            (
                debtor[0],
                creditor[0],
                payment
            )
        )

        debtor[1] -= payment
        creditor[1] -= payment

        if debtor[1] < 0.01:
            i += 1

        if creditor[1] < 0.01:
            j += 1

    print("\nSuggested Payments")

    if transactions:

        for debtor, creditor, amount in transactions:

            print(
                f"{debtor} should pay "
                f"₹{amount:.2f} to {creditor}"
            )

    else:
        print("No payments needed. Everyone is settled.")
```

```python
# Main Application
def expense_sharing_app():

    while True:

        print("\n" + "=" * 45)
        print("       EXPENSE SHARING APPLICATION")
        print("=" * 45)

        print("1. Add Expense")
        print("2. Calculate Settlements")
        print("3. Display Settlements")
        print("4. Suggest Payments")
        print("5. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":

            add_expense()

        elif choice == "2":

            balances = calculate_settlements()

            print("\nCalculated Balances:")

            for person, balance in balances.items():
                print(
                    f"{person}: ₹{balance:.2f}"
                )

        elif choice == "3":

            display_settlements()

        elif choice == "4":

            suggest_payment()

        elif choice == "5":

            print(
                "\nThank you for using "
                "Expense Sharing Application!"
            )

            break

        else:

            print("Invalid choice. Please try again.")
```

```python
# Run Application
expense_sharing_app()
```
